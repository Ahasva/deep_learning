```python
import numpy as np
```

### Sigmoid Function

```python
# Sigmoid activation function

def sigmoid(z):
    sigmoid = 1 / (1 + np.exp(-z))
    return sigmoid
```

### Sigmoid Derivative

```python
# This function determines how a (small) change in z2 affects y_hat

def sigmoid_derivative(sigmoid_output):
    sigmoid_derivative = sigmoid_output * (1 - sigmoid_output)
    # mathmatically the previous line correspondes to dy_hat_d_z2 = y_hat * (1 - y_hat)
    return sigmoid_derivative
```

### Variables for Testing of Artificial Neural Network

```python
#np.random.seed(42)  

# Randomly initializing the Target Value:
y = np.random.rand()

# Randomly initializing the raw inputs of the Input Layer:
x1, x2 = np.random.rand(2)

# Randomly initializing the Weights, the first two are applied to inputs of Input Layer
w1, w2, w3 = np.random.uniform(low=-1, high=1, size=3)

# Learning rate (often referred to as alpha or eta)
lr = 0.1

# Number of Epochs (each epoch refers to one complete pass through the entire training dataset)
epochs = 10000
# with several thousands of Epochs (around 10k) the Target Value (y)
# should be well approximated, i.e. approximately equals 100% accuracy
```

```python
print(f"INITIALIZATION:\n\tTarget Value:\n\t\t{y}\n\tInput Values:\n\t\tx1 = {x1};\n\t\tx2 = {x2}\n\tInitial Weights:\n\t\tw1 = {w1};\n\t\tw2 = {w2};\n\t\tw3 = {w3}")
```

### Building an Artificial Neural Network (with Python)

```python
for i in range(epochs):
    ### First neuron (Hidden Layer)
    # Step 1: linear combination
    z1 = x1 * w1 + x2 * w2
    # Step 2: nonlinear activation
    h1 = sigmoid(z1)

    ### Second neuron (Output Layer)
    # Step 1: Linear combination
    z2 = h1 * w3
    # Step 2: nonlinear activation
    y_hat = sigmoid(z2) # y_hat denotes the predicted value (at a certain time)

    ### Calculating the Mean Squared Error (MSE), i.e. the here used loss function
    MSE = 1/2 * (y - y_hat)**2

    ### Backpropagation, i.e. in each step you want to know how one element,
    # usually the Loss is effected by another element of the neural network)
    # Step 1: Partial derivative of the Loss with respect to y_hat
    dL_dy_hat = y_hat - y # reads: partial derivative (d) of the loss (L) with respect to the partial derivative of y (target value)
    # Step 2: Output neuron uses the sigmoid function (prep for step 3)
    dy_hat_dz2 = sigmoid_derivative(y_hat)
    # Step 3: Chain rule for the partial derivative of the Loss with respect to the partial derivative of z2
    dL_dz2 = dL_dy_hat * dy_hat_dz2
    # Step 4: Partial derivative of the Loss with respect to to w3
    dL_dw3 = dL_dz2 * h1 # h1 is equivalent to dz2_dw3 (the partial derivative of z2 with respect to the partial derivative of w3)
    # Step 5: Partial derivative of h1 with respect to z1 (prep for step 6)
    dh1_dz1 = sigmoid_derivative(h1) # mathmatically dh1_dz1 = h1 * (1 - h1)
    # Step 6: Partial derivative of the Loss with respect to the partial derivative of z1 (Chain Rule for dL_dz1)
    dL_dz1 = dL_dz2 * w3 * dh1_dz1
    # Step 7: Partial derivative of the Loss with respect to the partial derivatives of w1 and w2
    dL_dw1 = dL_dz1 * x1 # Mathmatically this corresponds to dL_dw1 = dL_dz1 * dz1_dw1 (dz1_dw1 = x1)
    dL_dw2 = dL_dz1 * x2 # Mathmatically this corresponds to dL_dw2 = dL_dz1 * dz1_dw2 (dz1_dw2 = x2)

    ### Gradient Descent
    # w1 <- w1 - alpha * dL_dw1
    w1_new = w1 - lr * dL_dw1
    w1 = w1_new

    # w2 <- w2 - alpha * dL_dw2
    w2_new = w2 - lr * dL_dw2
    w2 = w2_new
    
    # w3 <- w3 - alpha * dL_dw3
    w3_new = w3 - lr * dL_dw3
    w3 = w3_new

    if i % (epochs / 10) == 0 or i == epochs - 1:
        print(f"EPOCH NO. {i + 1}")
        print(f"\ty_hat:\t{y_hat:.17f};")
        print(f"\tMSE:\t{MSE:.17f};")
        print(f"\tw_new:\tw1 = {w1:.17f};\n\t\tw2 = {w2:.17f};\n\t\tw3 = {w3:.17f}\n")

print(f"SUMMARY:\n\tTraining of ANN in search for y = {y} completed after {epochs} Epochs.\n\tDelta of Target Value (y):\n\t\t{y}\n\tand the predicted value (y_hat):\n\t\t{y_hat}\n\tafter last Epoch:\n\t\t{np.abs(y_hat - y):.17f}.")
```
